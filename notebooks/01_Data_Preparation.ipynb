## Data Cleaning Assumptions

The following assumptions were made during the data-cleaning process:

### 1. Cabin

Missing values in the `Cabin` column were replaced with `"Unknown"`.

The value `"Unknown"` represents unavailable cabin information and does not mean that the passenger did not have a cabin.

### 2. Age

Missing values in the `Age` column were replaced with the median age of the dataset.

The median was selected because Age is a numerical variable and the median is less affected by extreme values than the mean.

### 3. Embarked

Missing values in the `Embarked` column were replaced with the most frequently occurring value (mode).

Only a small number of observations were missing, so using the most common category was considered reasonable.

### 4. Duplicate Records

The dataset was checked for duplicate rows. No duplicate rows were found, so no records were removed.

### 5. Data Types

The `Sex`, `Cabin`, and `Embarked` columns were converted to the `category` data type because they represent categorical information.

### 6. Data Preservation

No rows were removed during the cleaning process. Missing information was handled through appropriate replacement strategies wherever possible.

## Cleaning Summary

| Column | Problem | Action Taken |
|---|---|---|
| Cabin | Missing values | Replaced with `"Unknown"` |
| Age | Missing values | Replaced with median |
| Embarked | Missing values | Replaced with mode |
| All columns | Duplicate check | No duplicates found |
| Sex, Cabin, Embarked | Categorical data | Converted to `category` |

## Final Validation

The cleaned dataset was validated by checking:

- Data types
